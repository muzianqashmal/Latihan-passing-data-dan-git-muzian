```python
data = pd.read_excel(
    "muzian.xlsx",
    usecols="B:H",
    skiprows=6
)

data.columns = [
    "ID",
    "Name",
    "Quiz",
    "Mid_Term",
    "Final_Term",
    "Final_Score",
    "Grade"
]

display(data)
```

```python
print("Jumlah data:", len(data))
print("Jumlah kolom:", len(data.columns))
```

```python
print("Jumlah data:", len(data))
print("Jumlah kolom:", len(data.columns))

display(data)
```

```python
frekuensi = data["Grade"].value_counts().sort_index()

print("TABEL FREKUENSI GRADE")
print(frekuensi)
```

```python
tabel_frekuensi = pd.DataFrame({
    "Grade": frekuensi.index,
    "Frekuensi": frekuensi.values
})

display(tabel_frekuensi)
```

```python
persentase = (frekuensi / len(data)) * 100

tabel_frekuensi["Persentase (%)"] = persentase.round(2)

display(tabel_frekuensi)
```

```python
frekuensi.plot(
    kind="line",
    marker="o",
    figsize=(8, 5),
    title="Distribusi Grade Mahasiswa",
    xlabel="Grade",
    ylabel="Frekuensi"
);
```

```python
frekuensi.plot(
    kind="bar",
    figsize=(8, 5),
    title="Frekuensi Grade Mahasiswa",
    xlabel="Grade",
    ylabel="Jumlah Mahasiswa"
);
```

```python
frekuensi.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    title="Persentase Grade Mahasiswa",
    ylabel=""
);
```

```python
nilai = pd.to_numeric(data["Final_Score"], errors="coerce")

print("Data Final Score:")
print(nilai)
```

```python
mean = nilai.mean()
median = nilai.median()
modus = nilai.mode()

print("Mean   :", round(mean, 2))
print("Median :", round(median, 2))
print("Modus  :", modus.tolist())
```

```python
minimum = nilai.min()
maximum = nilai.max()

print("Minimum :", minimum)
print("Maximum :", maximum)
```

```python
Q1 = nilai.quantile(0.25)
Q2 = nilai.quantile(0.50)
Q3 = nilai.quantile(0.75)

print("Q1 :", round(Q1, 2))
print("Q2 :", round(Q2, 2))
print("Q3 :", round(Q3, 2))
```

```python
range_nilai = maximum - minimum

print("Range :", round(range_nilai, 2))
```

```python
variance = nilai.var()
std = nilai.std()

print("Variance :", round(variance, 2))
print("Standard Deviation :", round(std, 2))
```

```python
standard_error = std / (len(nilai) ** 0.5)

print("Standard Error :", round(standard_error, 2))
```

```python
skewness = nilai.skew()
kurtosis = nilai.kurt()

print("Skewness :", round(skewness, 3))
print("Kurtosis :", round(kurtosis, 3))
```

```python
IQR = Q3 - Q1

batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

print("IQR :", round(IQR, 2))
print("Batas bawah :", round(batas_bawah, 2))
print("Batas atas :", round(batas_atas, 2))
```

```python
outlier = nilai[
    (nilai < batas_bawah) |
    (nilai > batas_atas)
]

print("Data yang terindikasi outlier:")
print(outlier)
```
